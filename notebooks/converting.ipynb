# converting .arff to csv 

```python
from scipy.io import arff
```

```python
data, meta = arff.loadarff("../data/KC1.arff")
df = pd.DataFrame(data)
```

```python
df.to_csv("../data/KC1.csv", index=False)   #-------The '..' means "go up one folder"
```
